```python
#we should calculate wage, saving , cost_of_consumption(cost), funds he recieveid for r for cash flow


#wage
# Calculates the wage for the year n, if the person works for l(we suppose 30 years)  and his sallary will be increased at a rate of g_w percent.

def wage(n, l, g_w):
    # The person works for n years and the recieves a pension for l years and He will not recieve any money for his work during the retirement.
    if n <= l:
        return (1+g_w)**(n-1) 
    else :
        return 0
    
#saving
# Calculates the saving for year n, if the person works for l years and he will spend s percent of his sallary on savings.


def saving(n, l, s):
    if n <= l:
      return wage(n, l, 0.2) * s 
    else :
      return  0
    
#cost includes cost_of_consumption

# Calculate the amount of money spent in year n if the person works for l years, spent s percent of his sallary on savings. This is until the year l. Afterward,
# // he will recieve 80 percent of his lth sallary and this sallary will be increasedat a rate of g_p every year. .

def cost_of_consumption(n, l, s, g_p):
  
    if n <= l:
      return wage(n, l, 0.2) * (1 - s)  
    else:
      return wage(l, l, 0.2)*0.8*(1+g_p)**(n-(l+1))
    

#pension_fund
# Calculates the pension fund for the year n, if the person works for l years, and the pension fund invests the funds at a rate of r%. The pension fund has two entries each year:
#  1. The money spent on savings;
#  2. The return from investing the fund's ballance in previous year at an interest rate of r.


def pension_fund(n, l, r):
    sum = 0
    for i in range(1, n):
        sum += (wage(i, l, 0.2) - cost_of_consumption(i, l, 0.3, 0.2))*(1+r)**(n-i)
    if n > 0:
          return sum + wage(n, l, 0.2) - cost_of_consumption(n, l, 0.3, 0.2) 
    else :
           return 0


#cash_flow
# Calculates the cash flow of year n, if the person works for l years, and the pension fund invests the funds at a rate of r%. It computes just the flows in / out for every year. 
# The flow-in for every year (until n years) includes: 
# 1. The money spent on savings;
# 2. The return from investing the fund's ballance. The flow-out is after retirement and is 80 percent of the last working year's wage, increased at a rate of 20% every year.


def cashflow(n, l, r):
    
    if pension_fund(n-1, l, r) > 0:
        return_of_pension_fund = pension_fund(n-1, l, r) * r
    else:
        return_of_pension_fund = 0
    return wage(n, l, 0.2) - cost_of_consumption(n, l, 0.3, 0.2) + return_of_pension_fund
```

(ب) برای حل این سؤال، تابعی به نام npv(l, r, r_d) را تعریف می‌کنیم. این تابع NPV موجودی صندوق را در صورتی محاسبه می‌کند که فرد l سال کار می‌کند. نرخ بازده سرمایه‌گذاری صندوق r درصد است و نرخ تنزیل r_d است. این تابع برای سال‌های 1-58 ارزش فعلی جریان نقدی را محاسبه می‌کند.

```python
# The npv of pension fund calculated if the person will work for l years and then will recieve pension afterward. The sum of working and retirement years is 55.
# For each year, we calculate the present value of the year's cash flow discounted 
# at a rate of r_d%.
def npv(l, r, r_d):
    sum = 0
    for i in range(1, 58):
        sum += cashflow(i, l, r) / (1+r_d)**i
    return sum
npv(30, 0.17, 0.14)
```

در این قسمت تلاش کردیم تا نمودار npv  برای صندوق بازنشستی رسم کنیم.
 همانطور که در شکل دیده می‌شود حالت نزولی دارد ام در یکجا به پیک خود می‌رسدو

```python
import matplotlib.pyplot as plt

g_w = 0.2      
s = 0.3        
g_p = 0.2      
l = 30         
T = 57         
r = 0.17      
r_d = 0.14    

def wage(n):
    if n <= l:
        return (1 + g_w) ** (n - 1)
    else:
        return 0


def cost_of_consumption(n):
    if n <= l:
        return wage(n) * (1 - s)
    else:
        return wage(l) * 0.8 * (1 + g_p) ** (n - l - 1)


def pension_fund(n):
    fund = 0
    for i in range(1, n + 1):
        fund = fund * (1 + r) + wage(i) - cost_of_consumption(i)
    return fund


def cashflow(n):
    if n == 1:
        return wage(1) - cost_of_consumption(1)
    else:
        return wage(n) - cost_of_consumption(n) + pension_fund(n - 1) * r


def npv_up(T_end):
    value = 0
    for n in range(1, T_end + 1):
        value += cashflow(n) / (1 + r_d) ** n
    return value



years = list(range(1, T + 1))
npv_path = [npv_up(t) for t in years]



plt.figure()
plt.plot(years, npv_path)
plt.xlabel("Year")
plt.ylabel("Net Present Value")
plt.title("Life-Cycle NPV of Pension System")
plt.show()
```

```python
import matplotlib.pyplot as plt

def wage(n, l, g_w, initial_wage=1.0):
    if n <= l:
        return initial_wage * (1 + g_w)**(n - 1)
    else:
        return 0

def cost_of_consumption(n, l, s, g_p, g_w, initial_wage=1.0):
    if n <= l:
        return wage(n, l, g_w, initial_wage) * (1 - s)
    else:
        last_wage = wage(l, l, g_w, initial_wage)
        return last_wage * 0.8 * (1 + g_p)**(n - (l + 1))

def pension_fund(n, l, r, s, g_p, g_w, initial_wage=1.0):
    if n <= 0:
        return 0.0
    total = 0.0
    for i in range(1, n + 1):
        net_contribution = wage(i, l, g_w, initial_wage) * s if i <= l else -cost_of_consumption(i, l, s, g_p, g_w, initial_wage)
        total += net_contribution * (1 + r)**(n - i)
    return total


l = 30          
total_years = 57
s = 0.3        
g_w = 0.2       
g_p = 0.2       
r = 0.17        
initial_wage = 1.0

years = list(range(1, total_years + 1))

savings = [wage(n, l, g_w, initial_wage) * s if n <= l else 0.0 for n in years]
consumptions = [cost_of_consumption(n, l, s, g_p, g_w, initial_wage) for n in years]
pension_funds = [pension_fund(n, l, r, s, g_p, g_w, initial_wage) for n in years]


fig, axs = plt.subplots(3, 1, figsize=(12, 10))

axs[0].plot(years, savings, color='blue', linewidth=2)
axs[0].set_ylabel('Savings per Year')
axs[0].grid(True)
axs[0].axvline(x=l, color='gray', linestyle='--', label='Retirement')
axs[0].legend()

axs[1].plot(years, consumptions, color='orange', linewidth=2)
axs[1].set_ylabel('Consumption per Year')
axs[1].grid(True)
axs[1].axvline(x=l, color='gray', linestyle='--')

axs[2].plot(years, pension_funds, color='purple', linewidth=2)
axs[2].set_ylabel('Pension Fund Balance')
axs[2].set_xlabel('Year')
axs[2].grid(True)
axs[2].axvline(x=l, color='gray', linestyle='--')

plt.suptitle('NPV Analysis: Savings, Consumption, and Pension Fund Balance Over Time\n'
             f'(Wage growth {g_w*100}%, Savings {s*100}%, Return {r*100}%, Pension growth {g_p*100}%)')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


T = 60
years = np.arange(T)

# population
W0 = 25e6   # workers
R0 = 8e6    # retirees

# Growth rates brfore
g_w1 = 0.02
g_r1 = 0.05

# Growth rates (after )
g_w2 = 0.03
g_r2 = 0.02
policy_year = 20


avg_wage = 1.0
contribution_rate = 0.30
replacement_rate = 0.80

r = 0.17     
r_d = 0.14   
```

```python
#calculates the def population with its parameters mentioned
def population_paths(T, W0, R0, g_w1, g_r1, g_w2=None, g_r2=None, policy_year=None):
    W = np.zeros(T)
    R = np.zeros(T)

    W[0] = W0
    R[0] = R0

    for t in range(1, T):
        if policy_year is not None and t >= policy_year:
            W[t] = W[t-1] * (1 + g_w2)
            R[t] = R[t-1] * (1 + g_r2)
        else:
            W[t] = W[t-1] * (1 + g_w1)
            R[t] = R[t-1] * (1 + g_r1)

    return W, R

#calculates the def def pension_fund_flows with its parameters mentioned
def pension_fund_flows(W, R, avg_wage, c_rate, repl_rate, r):
    T = len(W)

    inflow = c_rate * avg_wage * W
    outflow = repl_rate * avg_wage * R

    cashflow = np.zeros(T)
    balance = np.zeros(T)

    for t in range(T):
        investment_return = r * balance[t-1] if t > 0 else 0
        cashflow[t] = inflow[t] - outflow[t] + investment_return
        balance[t] = (balance[t-1] if t > 0 else 0) + cashflow[t]

    return inflow, outflow, cashflow, balance

#calculate npv
def npv_fund(cashflow, r_d):
    return sum(cashflow[t] / (1 + r_d)**t for t in range(len(cashflow)))

```

```python
#calculation npv based population

W_base, R_base = population_paths(T, W0, R0, g_w1, g_r1)

inflow_b, outflow_b, cashflow_b, balance_b = pension_fund_flows(W_base, R_base, avg_wage, contribution_rate, replacement_rate, r)

npv_base = npv_fund(cashflow_b, r_d)
print("NPV :", round(npv_base, 2))
```

```python
#plotting

years = np.arange(len(inflow_b))


import matplotlib.pyplot as plt

fig, axs = plt.subplots(3, 1, figsize=(10, 8), sharex=True)

#  Inflows (Premiums) 
axs[0].bar(years, inflow_b / 1e6)
axs[0].set_title("Premium Inflows")
axs[0].set_ylabel("Millions")
axs[0].grid(axis="y")

#  Outflows (Pensions)
axs[1].bar(years, outflow_b / 1e6)
axs[1].set_title("Pension Outflows")
axs[1].set_ylabel("Millions")
axs[1].grid(axis="y")

#  Fund Balance 
axs[2].plot(years, balance_b / 1e6)
axs[2].axhline(0, color="black", linewidth=0.8)
axs[2].set_title("Pension Fund Balance")
axs[2].set_xlabel("Year")
axs[2].set_ylabel("Millions")
axs[2].grid(True)

plt.suptitle("Pension Fund Dynamics — Baseline Scenario", fontsize=14)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt


T = 60
years = np.arange(T)

# population
W0 = 25e6   # workers
R0 = 8e6    # retirees

# Growth rates 
g_w1 = 0.02
g_r1 = 0.05

# Growth rates after
g_w2 = 0.03
g_r2 = 0.02
policy_year = 20


avg_wage = 1.0
contribution_rate = 0.30
replacement_rate = 0.80

r = 0.17     
r_d = 0.14   

# definition of population with arguments described in the past

def population_paths(T, W0, R0, g_w1, g_r1, g_w2=None, g_r2=None, policy_year=None):
    W = np.zeros(T)
    R = np.zeros(T)

    W[0] = W0
    R[0] = R0

    for t in range(1, T):
        if policy_year is not None and t >= policy_year:
            W[t] = W[t-1] * (1 + g_w2)
            R[t] = R[t-1] * (1 + g_r2)
        else:
            W[t] = W[t-1] * (1 + g_w1)
            R[t] = R[t-1] * (1 + g_r1)

    return W, R
```

سخاوت بازنشستگی تأثیر قوی‌تری بر کسری بودجه نسبت به بازده سرمایه‌گذاری دارد.
عدم قطعیت سرمایه‌گذاری حتی زمانی که بازده مورد انتظار بالا باشد، ریسک ایجاد می‌کند.
سیاست‌های جمعیتی قابل اعتمادتر از مهندسی مالی صرف هستند.

```python
import numpy as np
import matplotlib.pyplot as plt

# Time 
T = 60
years = np.arange(T)

# Population
W0 = 25e6     # Initial workers
R0 = 8e6      # Initial retirees

# Growth rates
g_w = 0.02    
g_r = 0.05    

avg_wage = 1.0          
contribution_rate = 0.30
replacement_rate = 0.80


r = 0.20      # Mean investment return
r_d = 0.14   




```

```python
#Population Dynamics
def population_paths(T, W0, R0, g_w, g_r):
    W = np.zeros(T)
    R = np.zeros(T)
    W[0], R[0] = W0, R0

    for t in range(1, T):
        W[t] = W[t-1] * (1 + g_w)
        R[t] = R[t-1] * (1 + g_r)

    return W, R
#Pension Fund Accounting
def pension_fund(W, R, c_rate, repl_rate, r):
    inflow = c_rate * avg_wage * W
    outflow = repl_rate * avg_wage * R

    cashflow = np.zeros(len(W))
    balance = np.zeros(len(W))

    for t in range(len(W)):
        investment_return = r * balance[t-1] if t > 0 else 0
        cashflow[t] = inflow[t] - outflow[t] + investment_return
        balance[t] = (balance[t-1] if t > 0 else 0) + cashflow[t]

    return inflow, outflow, cashflow, balance
#Net Present Value

def npv(cashflow, r_d):
    return sum(cashflow[t] / (1 + r_d)**t for t in range(len(cashflow)))
```

```python
#POLICY ANALYSIS — INCREASING RETIREMENT AGE
W_base, R_base = population_paths(T, W0, R0, g_w, g_r)

#  retirement age increases → fewer retirees
g_r_policy = 0.035
W_pol, R_pol = population_paths(T, W0, R0, g_w, g_r_policy)


_, _, cf_base, _ = pension_fund(W_base, R_base, contribution_rate, replacement_rate, r)
_, _, cf_pol, _ = pension_fund(W_pol, R_pol, contribution_rate, replacement_rate, r)

print("NPV :", round(npv(cf_base, r_d), 2))
print("NPV with Higher Retirement Age:", round(npv(cf_pol, r_d), 2))
```

```python
#sensitivity
#Sensitivity to Investment Return 
r_values = [0.16, 0.20, 0.24]
npv_r = []

for r_test in r_values:
    _, _, cf, _ = pension_fund(W_base, R_base, contribution_rate, replacement_rate, r_test)
    npv_r.append(npv(cf, r_d))

#Sensitivity to Pension Growth 
repl_rates = [0.7, 0.8, 0.9]
npv_gp = []

for repl in repl_rates:
    _, _, cf, _ = pension_fund(W_base, R_base, contribution_rate, repl, r)
    npv_gp.append(npv(cf, r_d))


#plot


plt.subplot(1,2,1)
plt.plot(r_values, npv_r, marker='o')
plt.title("Sensitivity to Investment Return")
plt.xlabel("r")
plt.ylabel("NPV")

plt.subplot(1,2,2)
plt.plot(repl_rates, npv_gp, marker='o')
plt.title("Sensitivity to Pension Generosity")
plt.xlabel("Replacement Rate")
plt.ylabel("NPV")

plt.tight_layout()
plt.show()
    
```

```python
# Baseline population paths (already defined earlier)
W_base, R_base = population_paths(T, W0, R0, g_w, g_r)

# Baseline fund NPV
_, _, cf_base, _ = pension_fund(
    W_base, R_base,
    contribution_rate, replacement_rate, r
)
npv_base = npv(cf_base, r_d)


worker_shocks = [0.9, 1.0, 1.1]  # −10%, baseline, +10%
npv_workers = []


for shock in worker_shocks:
    W_shock = W_base * shock  # proportional change
    _, _, cf, _ = pension_fund(
        W_shock, R_base,
        contribution_rate, replacement_rate, r
    )
    npv_workers.append(npv(cf, r_d))
```

```python
plt.figure(figsize=(7,5))
plt.plot(worker_shocks, npv_workers, marker='o')
plt.axhline(npv_base, linestyle='--', label="Baseline NPV")
plt.title("Sensitivity to Number of Workers")
plt.xlabel("Worker Population Multiplier")
plt.ylabel("NPV")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Sensitivity to NUMBER OF RETIREES

retiree_shocks = [0.9, 1.0, 1.1]  
npv_retirees = []
  

#Compute NPVs

for shock in retiree_shocks:
    R_shock = R_base * shock
    _, _, cf, _ = pension_fund(
        W_base, R_shock,
        contribution_rate, replacement_rate, r
    )
    npv_retirees.append(npv(cf, r_d))

```

```python
#plot

plt.figure(figsize=(7,5))
plt.plot(retiree_shocks, npv_retirees, marker='o', color='red')
plt.axhline(npv_base, linestyle='--', label="Baseline NPV")
plt.title("Sensitivity to Number of Retirees")
plt.xlabel("Retiree Population Multiplier")
plt.ylabel("NPV")
plt.legend()
plt.grid(True)
plt.show()
```
